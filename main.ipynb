Importing the relevant libraries

```python
import numpy   as np
import pandas  as pd    
import matplotlib.pyplot as plt
import statsmodels.api as sm
import seaborn as sns
from sklearn.linear_model import LinearRegression    
sns.set()
```

Loading the raw data


```python
raw_data  =  pd.read_csv('file.csv')
raw_data.head()
```

**Preprocessing**


Exploring the descriptive statistics of the variables


```python
raw_data.describe(include = 'all')
```

Determinig the variables of interest

```python
data = raw_data.drop(['Model'], axis = 1)
data.describe(include = 'all')
```

Dealing with missing values

```python
data.isnull().sum() # sumof all missing values
```

```python
data_no_mv = data.dropna(axis = 0) # drop all rows with missing values
data_no_mv.describe(include = 'all')
```

Exploring the PDFs

```python
sns.displot(data_no_mv['Price'])
```

Dealing with outliers



```python
q = data_no_mv['Price'].quantile(0.99) # 99th percentile of Price  
data_1 = data_no_mv[data_no_mv['Price']<q] # remove all rows with Price > 99th percentile 
data_1.describe(include = 'all')    
```

```python
sns.displot(data_1['Price'])
```

```python
q = data_1['Mileage'].quantile(0.99) # 99th percentile of Mileage
data_2 = data_1[data_1['Mileage']<q] # remove all
```
